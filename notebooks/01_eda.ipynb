```python
import pandas as pd
```

```python
labeled = pd.read_parquet("../data/labels.parquet")
unk = pd.read_parquet("../data/tces_unk.parquet")

per_star = labeled.groupby("kepid")

n_stars = labeled["kepid"].nunique()
print(n_stars)
tces_per_star = per_star.size()
print(tces_per_star.value_counts().sort_index())

mixed = per_star["label"].nunique() > 1
print(mixed.sum())

star_label = per_star["label"].max()
print(star_label.mean())

overlap = set(labeled["kepid"]) & set(unk["kepid"])
print(len(overlap))
```
